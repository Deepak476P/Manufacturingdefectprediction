```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

from sklearn.pipeline import Pipeline

from sklearn.feature_selection import SelectKBest, mutual_info_classif

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)
```

```python
df = pd.read_csv("../data/ai4i2020.csv")

print("Dataset shape:", df.shape)
```

```python
columns_to_remove = [
    "UDI",
    "Product ID",
    "TWF",
    "HDF",
    "PWF",
    "OSF",
    "RNF"
]

df_clean = df.drop(columns=columns_to_remove)
```

```python
X = df_clean.drop(columns=["Machine failure"])
y = df_clean["Machine failure"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training:", X_train.shape)
print("Testing:", X_test.shape)
```

```python
categorical_features = ["Type"]

numerical_features = [
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]"
]
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numerical_features
        ),
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        )
    ]
)
```

```python
# Preprocess the training and testing data

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

# Get transformed feature names
feature_names = preprocessor.get_feature_names_out()

# Convert processed arrays into DataFrames
X_train_df = pd.DataFrame(
    X_train_processed,
    columns=feature_names,
    index=X_train.index
)

X_test_df = pd.DataFrame(
    X_test_processed,
    columns=feature_names,
    index=X_test.index
)

print("Training data:", X_train_df.shape)
print("Testing data:", X_test_df.shape)

X_train_df.head()
```

```python
logistic_model = Pipeline([
    ("preprocessing", preprocessor),
    ("classifier", LogisticRegression(
        max_iter=1000,
        class_weight="balanced"
    ))
])
```

```python
logistic_model.fit(X_train, y_train)
```

```python
y_pred = logistic_model.predict(X_test)
y_prob = logistic_model.predict_proba(X_test)[:, 1]
```

```python
print("Accuracy :", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall   :", recall_score(y_test, y_pred))
print("F1 Score :", f1_score(y_test, y_pred))
print("ROC-AUC  :", roc_auc_score(y_test, y_prob))
```

```python
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
decision_tree_model = Pipeline([
    ("preprocessing", preprocessor),
    ("classifier", DecisionTreeClassifier(
        max_depth=5,
        class_weight="balanced",
        random_state=42
    ))
])
```

```python
decision_tree_model.fit(X_train, y_train)
```

```python
dt_pred = decision_tree_model.predict(X_test)
dt_prob = decision_tree_model.predict_proba(X_test)[:, 1]
```

```python
print("Accuracy :", accuracy_score(y_test, dt_pred))
print("Precision:", precision_score(y_test, dt_pred))
print("Recall   :", recall_score(y_test, dt_pred))
print("F1 Score :", f1_score(y_test, dt_pred))
print("ROC-AUC  :", roc_auc_score(y_test, dt_prob))
```

```python
random_forest_model = Pipeline([
    ("preprocessing", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=200,
        max_depth=10,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])
```

```python
random_forest_model.fit(X_train, y_train)
```

```python
rf_pred = random_forest_model.predict(X_test)
rf_prob = random_forest_model.predict_proba(X_test)[:, 1]
```

```python
print("Accuracy :", accuracy_score(y_test, rf_pred))
print("Precision:", precision_score(y_test, rf_pred))
print("Recall   :", recall_score(y_test, rf_pred))
print("F1 Score :", f1_score(y_test, rf_pred))
print("ROC-AUC  :", roc_auc_score(y_test, rf_prob))
```

```python
results = []

models = {
    "Logistic Regression": (logistic_model, y_pred, y_prob),
    "Decision Tree": (decision_tree_model, dt_pred, dt_prob),
    "Random Forest": (random_forest_model, rf_pred, rf_prob)
}

for name, (model, predictions, probabilities) in models.items():

    results.append({
        "Model": name,
        "Feature Set": "All Features",
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision": precision_score(y_test, predictions),
        "Recall": recall_score(y_test, predictions),
        "F1 Score": f1_score(y_test, predictions),
        "ROC-AUC": roc_auc_score(y_test, probabilities)
    })

results_df = pd.DataFrame(results)

results_df
```

```python
from sklearn.feature_selection import SelectKBest, mutual_info_classif

selector_mi = SelectKBest(
    score_func=mutual_info_classif,
    k=5
)

X_train_selected = selector_mi.fit_transform(
    X_train_df,
    y_train
)

X_test_selected = selector_mi.transform(
    X_test_df
)

selected_features = feature_names[
    selector_mi.get_support()
]

print("Selected features:")
for feature in selected_features:
    print(feature)
```

```python
logistic_selected = LogisticRegression(
    max_iter=1000,
    class_weight="balanced"
)

logistic_selected.fit(
    X_train_selected,
    y_train
)

lr_selected_pred = logistic_selected.predict(
    X_test_selected
)

lr_selected_prob = logistic_selected.predict_proba(
    X_test_selected
)[:, 1]
```

```python
print("Accuracy :", accuracy_score(y_test, lr_selected_pred))
print("Precision:", precision_score(y_test, lr_selected_pred))
print("Recall   :", recall_score(y_test, lr_selected_pred))
print("F1 Score :", f1_score(y_test, lr_selected_pred))
print("ROC-AUC  :", roc_auc_score(y_test, lr_selected_prob))
```

```python
decision_tree_selected = DecisionTreeClassifier(
    max_depth=5,
    class_weight="balanced",
    random_state=42
)

decision_tree_selected.fit(
    X_train_selected,
    y_train
)

dt_selected_pred = decision_tree_selected.predict(
    X_test_selected
)

dt_selected_prob = decision_tree_selected.predict_proba(
    X_test_selected
)[:, 1]
```

```python
print("Accuracy :", accuracy_score(y_test, dt_selected_pred))
print("Precision:", precision_score(y_test, dt_selected_pred))
print("Recall   :", recall_score(y_test, dt_selected_pred))
print("F1 Score :", f1_score(y_test, dt_selected_pred))
print("ROC-AUC  :", roc_auc_score(y_test, dt_selected_prob))
```

```python
random_forest_selected = RandomForestClassifier(
    n_estimators=200,
    max_depth=10,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

random_forest_selected.fit(
    X_train_selected,
    y_train
)

rf_selected_pred = random_forest_selected.predict(
    X_test_selected
)

rf_selected_prob = random_forest_selected.predict_proba(
    X_test_selected
)[:, 1]
```

```python
print("Accuracy :", accuracy_score(y_test, rf_selected_pred))
print("Precision:", precision_score(y_test, rf_selected_pred))
print("Recall   :", recall_score(y_test, rf_selected_pred))
print("F1 Score :", f1_score(y_test, rf_selected_pred))
print("ROC-AUC  :", roc_auc_score(y_test, rf_selected_prob))
```

```python
selected_results = [
    {
        "Model": "Logistic Regression",
        "Feature Set": "MI Selected (5)",
        "Accuracy": accuracy_score(y_test, lr_selected_pred),
        "Precision": precision_score(y_test, lr_selected_pred),
        "Recall": recall_score(y_test, lr_selected_pred),
        "F1 Score": f1_score(y_test, lr_selected_pred),
        "ROC-AUC": roc_auc_score(y_test, lr_selected_prob)
    },
    {
        "Model": "Decision Tree",
        "Feature Set": "MI Selected (5)",
        "Accuracy": accuracy_score(y_test, dt_selected_pred),
        "Precision": precision_score(y_test, dt_selected_pred),
        "Recall": recall_score(y_test, dt_selected_pred),
        "F1 Score": f1_score(y_test, dt_selected_pred),
        "ROC-AUC": roc_auc_score(y_test, dt_selected_prob)
    },
    {
        "Model": "Random Forest",
        "Feature Set": "MI Selected (5)",
        "Accuracy": accuracy_score(y_test, rf_selected_pred),
        "Precision": precision_score(y_test, rf_selected_pred),
        "Recall": recall_score(y_test, rf_selected_pred),
        "F1 Score": f1_score(y_test, rf_selected_pred),
        "ROC-AUC": roc_auc_score(y_test, rf_selected_prob)
    }
]

selected_results_df = pd.DataFrame(selected_results)

selected_results_df
```

```python
final_comparison = pd.concat(
    [results_df, selected_results_df],
    ignore_index=True
)

final_comparison
```

```python
from sklearn.feature_selection import RFE
```

```python
rfe_estimator = LogisticRegression(
    max_iter=1000,
    class_weight="balanced"
)

rfe_selector = RFE(
    estimator=rfe_estimator,
    n_features_to_select=5
)
```

```python
rfe_selector.fit(
    X_train_df,
    y_train
)
```

```python
rfe_selected_features = feature_names[
    rfe_selector.support_
]

print("RFE Selected Features:")

for feature in rfe_selected_features:
    print(feature)
```

```python
X_train_rfe = rfe_selector.transform(X_train_df)
X_test_rfe = rfe_selector.transform(X_test_df)

print("RFE training shape:", X_train_rfe.shape)
print("RFE testing shape:", X_test_rfe.shape)
```

```python
dt_rfe = DecisionTreeClassifier(
    max_depth=5,
    class_weight="balanced",
    random_state=42
)

dt_rfe.fit(X_train_rfe, y_train)

dt_rfe_pred = dt_rfe.predict(X_test_rfe)
dt_rfe_prob = dt_rfe.predict_proba(X_test_rfe)[:, 1]
```

```python
dt_rfe_metrics = {
    "Model": "Decision Tree",
    "Feature Set": "RFE Selected (5)",
    "Accuracy": accuracy_score(y_test, dt_rfe_pred),
    "Precision": precision_score(y_test, dt_rfe_pred),
    "Recall": recall_score(y_test, dt_rfe_pred),
    "F1 Score": f1_score(y_test, dt_rfe_pred),
    "ROC-AUC": roc_auc_score(y_test, dt_rfe_prob)
}

dt_rfe_metrics
```

```python
rfe_results_df = pd.DataFrame([
    lr_rfe_metrics,
    dt_rfe_metrics,
    rf_rfe_metrics
])

complete_results = pd.concat(
    [
        results_df,
        selected_results_df,
        rfe_results_df
    ],
    ignore_index=True
)

complete_results
```
